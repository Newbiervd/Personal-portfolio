Lab 6.1:
* Simple Calculator Functions
Objective: Students will define and use functions to perform basic arithmetic operations, practicing parameter passing and return values.

Tasks (Common for Local & Colab):
Create a new Python file (lab6_1.py) or Colab code cell.Define four functions:
 * add(a, b), subtract(a, b), multiply(a, b), divide(a, b).
Each function should take two numbers as parameters.
Each function should return the result of the operation.
Add a docstring to each function explaining its purpose.

```python
# Lab 6.1
def add(a, b):
    """Perform basic addition."""
    return a + b

def subtract(a, b):
    """Perform basic subtraction."""
    return a - b

def multiply(a, b):
    """Perform basic multiplication."""
    return a * b

def divide(a, b):
    """Perform basic division and handle zero division error."""
    if b == 0:
        return "Error: Division by zero"
    return a / b

def power(base, exponent=2):
    """Calculate power with a default exponent of 2."""
    return base ** exponent

while True:
    print("\n--- Simple Calculator ---")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Power")
    print("6. Exit")

    choice = input("Choose an operation (1-6): ")

    if choice == '6':
        break

    if choice in ['1', '2', '3', '4', '5']:
        try:
            num1 = float(input("Enter first number: "))

            if choice == '5':
                exp_input = input("Enter exponent (Press Enter for square): ")
                if exp_input.strip() == "":
                    result = power(num1)
                else:
                    result = power(num1, float(exp_input))
            else:
                num2 = float(input("Enter second number: "))
                if choice == '1':
                    result = add(num1, num2)
                elif choice == '2':
                    result = subtract(num1, num2)
                elif choice == '3':
                    result = multiply(num1, num2)
                elif choice == '4':
                    result = divide(num1, num2)

            print(f"Result: {result}")
        except ValueError:
            print("Invalid input. Please enter numbers.")
    else:
        print("Invalid choice.")
```

Objective:

Tasks (Common for Local & Colab):

Part 2: Use the Custom Module and Standard Modules:
* Create another new Python file (main_program.py) or Colab code cell.
* At the top of main_program.py, import your my_utils module: import my_utils.
* Import standard modules: import math, import random.

* Test my_utils.is_prime() with a few numbers (e.g., 7, 10).
* Use math.sqrt() to calculate the square root of a number.
* Use random.randint() to generate a random number between 1 and 100.
* For Colab specifics: If my_utils.py is not easily managed as a separate file, the main_program cell can simply define the functions from my_utils within itself or the instructor can pre-upload my_utils.py to Colab. The best way for true module demo in Colab is to use %%writefile my_utils.py magic command in one cell to create the file, then import it in another.

```python
# my_utils
def greet(name):
    """Print a greeting message."""
    print(f"Hello, {name}!")

def is_prime(number):
    """Return True if number is prime, False otherwise."""
    if number <= 1:
        return False
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return False
    return True
```

```python
# main Program
import my_utils
import math
import random

my_utils.greet("Alice")

print(f"Is 7 prime?: {my_utils.is_prime(7)}")
print(f"Is 10 prime?: {my_utils.is_prime(10)}")

print(f"Square root of 64: {math.sqrt(64)}")

print(f"Random number (1-100): {random.randint(1, 100)}")
```

```python
def check_sign(number):
    """Determine if a number is positive, negative, or zero."""
    if number > 0:
        return "positive"
    elif number < 0:
        return "negative"
    else:
        return "zero"

def check_parity(number):
    """Determine if a number is even or odd."""
    if number % 2 == 0:
        return "even"
    else:
        return "odd"

try:
    user_num = int(input("Enter an integer: "))
    sign = check_sign(user_num)
    parity = check_parity(user_num)
    print(f"The number is {sign} and {parity}.")
except ValueError:
    print("Please enter a valid integer.")
```
