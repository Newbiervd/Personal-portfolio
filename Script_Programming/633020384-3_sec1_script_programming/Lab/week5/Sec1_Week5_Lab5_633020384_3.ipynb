## 🔄 ส่วนที่ 1: Week 4 Recap (5-Min Review)
### 📖 สรุปองค์ความรู้สัปดาห์ที่ 4
ในสัปดาห์ที่ผ่านมา เราได้ศึกษาโครงสร้างข้อมูลแบบลำดับ (Sequence) สองรูปแบบหลักที่มีบทบาทสำคัญในการจัดระเบียบข้อมูล:
1. **Lists (`list`):** โครงสร้างแบบเรียงลำดับที่สามารถเปลี่ยนแปลงค่าได้ (**Mutable**) รองรับการเพิ่มข้อมูลด้วย `.append()`, ขยายด้วย `.extend()`, แทรกด้วย `.insert()`, และลบด้วย `.remove()` หรือ `.pop()` เหมาะสำหรับข้อมูลที่มีการปรับปรุงหรือเปลี่ยนแปลงตลอดเวลา
2. **Tuples (`tuple`):** โครงสร้างแบบเรียงลำดับที่ไม่สามารถเปลี่ยนแปลงค่าได้ (**Immutable**) ทำหน้าที่เป็นเกราะป้องกันข้อมูลจากการถูกแก้ไขโดยระบายภัยหรือไม่ได้ตั้งใจ เช่น พิกัดภูมิศาสตร์ `(latitude, longitude)` หรือค่าสี RGB `(255, 255, 255)` ซึ่งหากเราพยายามแก้ไข สัญกรณ์ของ Python จะฟ้องข้อผิดพลาด `TypeError` ทันทีเพื่อรักษาความถูกต้องและความปลอดภัยของข้อมูล (**Data Integrity**)

```python
# === ทบทวนความแตกต่างระหว่าง List (Mutable) และ Tuple (Immutable) ===
student_list = ["Alice", "Bob", "Charlie", "Howl"]   # List สามารถเพิ่ม/แก้ไขได้
student_list.append("David")
print(f"Updated List: {student_list}")

security_config = ("192.168.1.1", 8080)        # Tuple คงที่ ห้ามแก้ไข
print(f"Secure Connection Config: {security_config}")

try:
    # พยายามแก้ไข Tuple จะทำให้ระบบแจ้งเตือนข้อผิดพลาดทันที
    security_config[0] = "10.0.0.1"
except TypeError as e:
    print(f"🚨 Alert! [Data Integrity Safeguard Activated]: {e}")
```

### 1. โครงสร้างข้อมูลประเภทดิกชันนารี (Dictionaries - `dict`)
ดิกชันนารีคือโครงสร้างข้อมูลประเภท **Mapping** ที่จับคู่กันระหว่าง **Key** และ **Value** โดยมีกฎสำคัญคือ:
* **Keys** จะต้องเป็นวัตถุที่ไม่สามารถเปลี่ยนแปลงค่าได้ (**Immutable** และ **Hashable**) เสมอ เช่น สตริง ตัวเลข หรือทูเพิล
* **Values** สามารถเป็นข้อมูลประเภทใดก็ได้ ไม่ว่าจะเป็นข้อมูลพื้นฐาน ลิสต์ หรือแม้กระทั่งดิกชันนารีซ้อนดิกชันนารี (Nested Dictionary)
* ตั้งแต่ Python 3.7 เป็นต้นมา ดิกชันนารีจะจดจำลำดับของการแทรกข้อมูลโดยปริยาย (Insertion Order Maintained)

#### 🛠️ เมธอดและตัวดำเนินการที่สำคัญ:
* **การประกาศสร้าง:** ใช้ปีกกา `{}` หรือฟังก์ชันตัวสร้าง `dict()`
* **การเข้าถึงแบบปลอดภัย:** เมธอด `.get(key, default)` ช่วยไม่ให้เกิดข้อผิดพลาดรุนแรง หากไม่พบคีย์จะส่งกลับค่าเริ่มต้นที่ตั้งไว้แทนการแครชของแอปพลิเคชัน
* **การลบข้อมูล:** เมธอด `.pop(key)` ลบคีย์และส่งคืนค่าตัวแปรนั้น เมธอด `.popitem()` ลบข้อมูลตัวสุดท้าย และคำสั่งลบแบบดั้งเดิมอย่าง `del`
* **ตัวดำเนินการผสาน (Python 3.9+):** เครื่องหมาย `|` ใช้ผสานสองดิกชันนารีเข้าด้วยกันโดยสร้างวัตถุใหม่ และ `|=` ใช้ผสานปรับปรุงข้อมูลในตำแหน่งเดิม (In-place update)

```python
# === การทดลองการใช้งาน Dictionary ขั้นพื้นฐานและขั้นสูง ===

# 1. สร้าง Dictionary เก็บข้อมูลโปรไฟล์ผู้ใช้งาน
user_profile = {
    "username": "admin_secure",
    "role": "security_officer",
    "access_level": 5
    }
print(f"Initial Profile: {user_profile}")

# 2. การเข้าถึงข้อมูลแบบปกติ vs แบบปลอดภัย (.get)
print(f"Direct Access [Role]: {user_profile['role']}")
print(f"Safe Access [Last Login]: {user_profile.get('last_login', 'Never Logged In')}") # คีย์ไม่มี แต่ระบบไม่พัง
#print(f"Direct Access [Role]: {user_profile['ip']}")

# 3. การเพิ่มและแก้ไขข้อมูล
user_profile["status"] = "Active"          # เพิ่มคีย์ใหม่
user_profile["access_level"] = 6            # ปรับปรุงค่าคีย์เดิม
print(f"Added and Updated Profile: {user_profile}")

# 4. การทดลองใช้ Modern Merge Operator (| และ |=) ตามมาตรฐาน Python 3.9+
extra_info = {"location": "Bangkok Center", "status": "Verified"}
combined_profile = user_profile | extra_info # ข้อมูลทางขวา (extra_info) จะเขียนทับคีย์ที่ซ้ำกัน
print(f"Combined Profile via (|): {combined_profile}")

# 5. การนำข้อมูลออกด้วย .pop()
removed_status = combined_profile.pop("status")
print(f"Removed Status Value: {removed_status}")
print(f"Final Profile State: {combined_profile}")
```

### 2. โครงสร้างข้อมูลประเภทเซต (Sets - `set`)
เซตคือโครงสร้างข้อมูลที่เป็น **Unordered** (ไม่มีการเรียงลำดับ) และสมาชิกทุกตัวต้อง **Unique** (ห้ามซ้ำกันโดยเด็ดขาด) นิยมนำมาใช้ในการล้างข้อมูลที่ซ้ำซ้อนและการประมวลผลเซตทางคณิตศาสตร์

#### 🛠️ เมธอดและการจัดการความปลอดภัยทางไวยากรณ์:
* **การนำข้อมูลออกอย่างปลอดภัย:** เมื่อต้องการลบข้อมูลออกจากเซต เมธอด `.remove(value)` จะสร้างข้อยกเว้นข้อผิดพลาดแครชโปรแกรมทันทีหากไม่พบค่าดังกล่าว ในขณะที่เมธอด **`.discard(value)`** จะทำงานอย่างปลอดภัย หากไม่พบค่าจะข้ามผ่านทันทีโดยไม่มีข้อผิดพลาดรุนแรง
* **ตัวดำเนินการคณิตศาสตร์เชิงเซต:**
  * **Union (`|` หรือ `.union()`):** รวมสมาชิกทั้งหมดจากทั้งสองเซต
  * **Intersection (`&` หรือ `.intersection()`):** เลือกเอาเฉพาะสมาชิกที่มีร่วมกันทั้งสองเซต
  * **Difference (`-` หรือ `.difference()`):** สมาชิกที่อยู่เซตแรก แต่ไม่อยู่ในเซตหลัง
  * **Symmetric Difference (`^` หรือ `.symmetric_difference()`):** สมาชิกที่อยู่เซตใดเซตหนึ่งแต่ไม่ได้อยู่ร่วมกันทั้งสองเซต

  (set1 union set2) - (set1 intersect set2)

```python
# === การทดลองการใช้งาน Set และการดำเนินการทางคณิตศาสตร์เชิงเซต ===

# 1. สร้างเซตข้อมูลสกิลของพนักงานไอที
my_skills = {"Python", "SQL", "Git", "Cloud"}  # = {"Cloud", "Python", "SQL", "Git"}
required_skills = {"Python", "SQL", "Java"}

print(f"My Skills: {my_skills}")
print(f"Required Position Skills: {required_skills}")

# 2. ดำเนินการทางคณิตศาสตร์เชิงเซต (Set Operators)
print(f"Union (ทักษะรวมทั้งหมด): {my_skills | required_skills}")
print(f"Intersection (ทักษะที่เรามีตรงกับความต้องการ): {my_skills & required_skills}")
print(f"Difference (ทักษะที่เรายังขาดอยู่): {required_skills - my_skills}")
print(f"Difference 2 (ทักษะที่เรายังขาดอยู่): {my_skills - required_skills}")
print(f"Symmetric Difference (ทักษะที่ไม่ซ้ำซ้อนกัน): {my_skills ^ required_skills}")
print(f"Symmetric Difference 2 (ทักษะที่ไม่ซ้ำซ้อนกัน): {required_skills ^ my_skills}")

# 3. เปรียบเทียบความปลอดภัยระหว่าง .remove() และ .discard()
print("\n--- Testing Set Element Removal Safety ---")
my_skills.discard("C++") # คีย์ไม่มีในเซต แต่โปรแกรมรันต่อไปได้อย่างเสถียรไร้รอยต่อ
print("Using .discard('C++') -> Safe execution completed.")

try:
    my_skills.remove("C++") # จะเกิดข้อยกเว้นและหยุดการทำงานทันทีหากไม่มี try-except
except KeyError as e:
    print(f"Using .remove('C++') -> 🚨 Raised Exception: Key {e} not found in set!")
```

### 🛠️ Lab 5.1: ระบบค้นหาทะเบียนข้อมูลนักศึกษา (Student Registry System via Dictionary)
**วัตถุประสงค์:** ให้นักศึกษาฝึกฝนการใช้โครงสร้างข้อมูลประเภท Dictionary ในการเก็บข้อมูลโปรไฟล์และใช้วิธีสืบค้นอย่างปลอดภัยระดับ $O(1)$

**โจทย์ภารกิจ:**
1. สร้างฐานข้อมูลนักศึกษาจำลองในรูปของดิกชันนารีซ้อนดิกชันนารี (Nested Dictionary) ประกอบด้วยรหัสนักศึกษา ชื่อ และเกรดเฉลี่ย
2. เขียนระบบรับค่า Input จากยูสเซอร์ป้อนรหัสนักศึกษาเพื่อค้นหาข้อมูล
3. นำเมธอด `.get()` มาปรับใช้เพื่อแสดงข้อความแจ้งเตือนอย่างปลอดภัยเมื่อไม่พบข้อมูลในระบบ ป้องกันไม่ให้โปรแกรมแครช

```python
# === เขียนโค้ดส่งงาน Lab 5.1 ด้านล่างนี้ ===
student_db = {
    "6901001": {"name": "Somchai", "gpa": 3.75},
    "6901002": {"name": "Somsri", "gpa": 3.20},
    "6901003": {"name": "Somsak", "gpa": 3.90}
}

# จำลองการสืบค้นข้อมูล
search_id = input("กรุณาป้อนรหัสนักศึกษาที่ต้องการค้นหา: ")
result = student_db.get(search_id, "⚠️ ไม่พบข้อมูลรหัสนักศึกษาดังกล่าวในระบบระเบียน")

print(f"ผลการสืบค้นข้อมูล: {result}")
```

### 🛠️ Lab 5.2: ระบบคัดกรองข้อมูลประวัติทำความสะอาดข้อมูลซ้ำซ้อน (Data De-duplication & Skill Audit via Set)
**วัตถุประสงค์:** ประยุกต์ใช้เซตในการคัดกรองตัวเลขซ้ำและตรวจเช็กเปรียบเทียบชุดทักษะเพื่อหาช่องว่างขีดความสามารถ (Skill Gap Analysis)

**โจทย์ภารกิจ:**
1. มีลิสต์ประวัติหมายเลขไอพีแอดเดรสที่เข้ามาติดต่อระบบเซิร์ฟเวอร์ที่มีข้อมูลซ้ำซ้อนกันจำนวนมาก ให้นักศึกษาแปลงร่างข้อมูลตัวนี้ให้กลายเป็นเซตเพื่อหาที่อยู่ไอพีที่ไม่ซ้ำกันทั้งหมด (Unique IPs)
2. ทำการวิเคราะห์เปรียบเทียบเซตทักษะของพนักงานกับข้อกำหนดความปลอดภัยระบบไอทีเพื่อหาว่าพนักงานยังขาดทักษะด้านใดบ้างโดยใช้ตัวดำเนินการลบเซต (`-`)

```python
# === เขียนโค้ดส่งงาน Lab 5.2 ด้านล่างนี้ ===
raw_ip_logs = ["192.168.1.1", "10.0.0.2", "192.168.1.1", "172.16.0.5", "10.0.0.2"]

# 1. ล้างค่าไอพีที่ซ้ำซ้อน
unique_ips = set(raw_ip_logs)
print(f"จำนวนรายการไอพีที่ไม่ซ้ำกันในระบบ: {unique_ips}")

# 2. หาช่องว่างทักษะไอทีที่จำเป็น
comp_required = {"Firewall", "Linux", "Python", "Cryptography"}
emp_current = {"Python", "Linux"}
missing_skills = comp_required - emp_current

print(f"🚨 สกิลความปลอดภัยระบบไอทีที่พนักงานต้องเข้ารับการอบรมเพิ่ม: {missing_skills}")
```
